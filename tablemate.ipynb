```python
import csv #import csv från det lokala biblioteket sen öppnar filen tables.csv


def load_tables(file_import): #skapar en funktion --load_tables-- som läser in tabeller från en CSV-fil
    tables = [] #skapar listan tables som kommer att innehålla alla tabeller


    try:
        with open(file_import, "r", encoding="utf-8", newline="") as file: #öppnar datafilen
            reader = csv.DictReader(file) #använder rubriker som nycklar för att läsa in data
            for row in reader: #loopar igenom ett bord i taget
                row["table_id"] = int (row["table_id"]) #konverterar table_id till en int
                row["table_capacity"] = int (row["table_capacity"]) #konverterar table_capacity till en int
                row["is_bookable"] = row["is_bookable"].lower() == "true" #konverterar is_bookable till en bool
                tables.append(row) #lägger till raden i listan tables
        return tables

    except FileNotFoundError: #om filen inte hittas
        print(f"Error: The file '{file_import}' was not found.") #skriver ut ett felmeddelande
    
    except (ValueError, KeyError) as e: #om det finns ett värdefel eller nyckelfel
        print(f"Error: An error occurred while reading the file '{file_import}': {e}") #skriver ut ett felmeddelande
    except csv.Error as e: #om det finns ett csv-fel
        print(f"Error: An error occurred while processing the CSV file '{file_import}': {e}") #skriver ut ett felmeddelande
    return []


all_tables = load_tables("tables.csv") #anropar funktionen load_tables och lagrar resultatet i variabeln all_tables

print(f"Number of tables: {len(all_tables)}") #skriver ut antalet tabeller som har lästs in

for table in all_tables: #loopar igenom alla tabeller och skriver ut dem
    print(table)
```
